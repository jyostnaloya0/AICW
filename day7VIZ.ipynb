```python
import pandas as pd
```

```python
df=pd.read_csv("IO-DATA.csv")
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.barplot(data=df,x="gender",y="claim")
plt.title("BARPLOT")
plt.xlabel("GENDER")
plt.ylabel("CLAIM")
plt.show()
```

```python
sns.countplot(data=df,x="diabetic")
plt.show()
```

```python
sns.histplot(data=df,x="bmi",bins=6,kde=True)
plt.show()
```

```python
sns.kdeplot(data=df,x="bmi",fill=True)
plt.show()
```

```python
sns.boxplot(data=df,y="bmi")
plt.show()
```

```python
sns.boxplot(data=df,x="gender",y="bmi")
plt.show()
```

```python
sns.violinplot(data=df,x="region",y="claim")
plt.show()
```

```python
sns.stripplot(data=df,x="gender",y="claim")
plt.show()
```

```python
sns.swarmplot(data=df,x="gender",y="claim")
plt.show()
```

```python
sns.scatterplot(data=df,x="age",y="bmi")
plt.show()
```

```python
sns.scatterplot(data=df,x="age",y="bmi",hue="gender")
plt.show()
```
